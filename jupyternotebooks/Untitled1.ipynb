```python
from geof_interp_widget import (
    ESCALAS, build_selector_widget, interpolate_mosaic_and_update,
    export_geotiff_stack, save_all_stacks, run_workflow_quick
)

# (A) Interface com widgets:
widgets = build_selector_widget()  # carrega dados via botão; variable global 'quadricula' é atualizada
```

```python
# (B) Fluxo rápido sem widgets:
mc, quadricula = run_workflow_quick(
    escala="25k",
    ids=["SF23_YB_I1_NW"],
    gama="gama_line_1105", mag="mag_line_1105", extend_size=600
)


# Exporta 1 GeoTIFF por folha (bandas nas ordem de 'canais'):
save_all_stacks(quadricula, canais=["gama_line_1105","mag_line_1105"], outdir="out_tif")
```

```python
# Interpola no mosaico (com buffer) e recorta por folha (sem costura):
value_fields = { "gama_line_1105": "TC", "mag_line_1105": "TMI" }
quadricula = interpolate_mosaic_and_update(
    quadricula, value_fields, folhas_gdf=mc, epsg=mc.crs.to_epsg() if mc.crs else None,
    spacing=100.0, reduce_spacing=500.0, buffer_m=600.0, dtype="float32", nodata=float("nan")
)
```

```python
from geof_interp_widget import report_interpolation_status, save_all_stacks

# Conferir se cada folha recebeu as bandas interpoladas:
report_interpolation_status(quadricula, ["gama_line_1105","mag_line_1105"])

# Exportar apenas as folhas que estiverem completas:
save_all_stacks(quadricula, canais=["gama_line_1105","mag_line_1105"], outdir="out_tif", skip_missing=True, verbose=True)
```
