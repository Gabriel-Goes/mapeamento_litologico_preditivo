```python
from shapely.geometry import shape
import pandas as pd
from db_conn import get_folha_geom_geojson as folha
# alvo em UTC
target = pd.to_datetime("2008-06-15", utc=True)
bbox=shape(folha('SB21_ZA_II1_NE')).bounds
bbox
```

```python
import earthaccess
tspan = ("2004-06-01", "2012-06-30")

results_l1t = earthaccess.search_data(
    short_name="AST_L1T",
    version="004",
    cloud_hosted=True,          # força coleção no Earthdata Cloud
    temporal=tspan,
    bounding_box=bbox,
    count=25,                   # limite de granules
)

print("Granules AST_L1T.004 encontrados:", len(results_l1t))
print(results_l1t[0])
```

```python
collection_id_l1t = "C3306888411-LPCLOUD"  # AST_L1T.004

results_l1t = earthaccess.search_data(
    concept_id=collection_id_l1t,
    cloud_hosted=True,
    temporal=tspan,
    bounding_box=bbox,
    count=50,
)
```

```python
paths_l1t = earthaccess.download(results_l1t[:2], "./aster_l1t_004")
print(paths_l1t)
```

```python
files_l1t = earthaccess.open(results_l1t[:2])
# depois abrir com rasterio/rioxarray/xarray, dependendo do formato
```
