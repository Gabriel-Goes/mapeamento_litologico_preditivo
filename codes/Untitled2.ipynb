```python
# -*- coding: utf-8 -*-
from __future__ import annotations

import re
import tempfile
from pathlib import Path

import earthaccess as ea
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
import rioxarray as rxr
from shapely.geometry import shape, Polygon, MultiPolygon

from pystac_client import Client
import planetary_computer as pc


# ======================================================================
# 1) AST_07XT – nuvem, ordenação, preview SWIR B04
# ======================================================================

def _extract_numbers_from_attr(attr):
    vals = attr.get("Values") or []
    text = " ".join(str(v) for v in vals)
    nums_str = re.findall(r"[-+]?\d*\.?\d+(?:[eE][-+]?\d+)?", text)
    if not nums_str:
        return None
    return [float(s) for s in nums_str]


def get_cloud_from_umm(granule):
    umm = granule["umm"]
    attrs = umm.get("AdditionalAttributes", [])
    best = None
    for attr in attrs:
        name = (attr.get("Name") or "").lower()
        if "cloud" not in name:
            continue
        nums = _extract_numbers_from_attr(attr)
        if not nums:
            continue
        mean_val = float(np.mean(nums))
        if best is None:
            best = (mean_val, nums, name)
        else:
            if "cloud_coverage_data" in name and "cloud_coverage_data" not in best[2]:
                best = (mean_val, nums, name)
            elif mean_val < best[0]:
                best = (mean_val, nums, name)
    if best is None:
        return None, None, None
    return best


def sort_by_cloud_ast(results):
    tmp = []
    for g in results:
        cc_mean, cc_list, attr_name = get_cloud_from_umm(g)
        if cc_mean is None:
            cc_mean = 999.0
            attr_name = None
        tmp.append((cc_mean, g, attr_name))
    return sorted(tmp, key=lambda t: t[0])


def _get_granule_footprint(granule):
    """
    Constrói o footprint real do AST_07XT a partir de umm->SpatialExtent->GPolygons.
    """
    umm = granule["umm"]
    spatial = umm.get("SpatialExtent") or umm.get("SpatialCoverage") or {}
    hs = spatial.get("HorizontalSpatialDomain", {})
    geom = hs.get("Geometry", {})

    polys = []

    gpolys = geom.get("GPolygons") or []
    for gp in gpolys:
        pts = gp.get("Boundary", {}).get("Points", [])
        if not pts:
            continue
        coords = [(p["Longitude"], p["Latitude"]) for p in pts]
        if coords[0] != coords[-1]:
            coords.append(coords[0])
        polys.append(Polygon(coords))

    if not polys:
        return None
    if len(polys) == 1:
        return polys[0]
    return MultiPolygon(polys)


def _has_swir_b04(granule) -> bool:
    """
    Verifica se o granule possui banda SRF_SWIR_B04 nos data_links.
    """
    try:
        links = granule.data_links()
    except Exception:
        return False
    for url in links:
        if "SRF_SWIR_B04" in url:
            return True
    return False


def download_swir_b04(granule, out_dir=None):
    """
    Faz download de TODOS os arquivos do granule e retorna o Path da SRF_SWIR_B04.
    Assume que já filtramos somente granules que possuem essa banda.
    """
    out_dir = Path(out_dir or tempfile.mkdtemp(prefix="AST_07XT_"))
    files = ea.download([granule], local_path=out_dir)
    b4_list = [Path(f) for f in files if "SRF_SWIR_B04" in str(f)]
    if not b4_list:
        # segurança extra caso algum granule "passe" sem B04
        raise RuntimeError("Nenhum arquivo SRF_SWIR_B04* encontrado nesse granule.")
    return b4_list[0]


def clip_to_geom(tif_path, geom_geojson):
    g = shape(geom_geojson)
    gdf = gpd.GeoDataFrame(geometry=[g], crs="EPSG:4326")
    da = rxr.open_rasterio(tif_path, masked=True)
    gdf_proj = gdf.to_crs(da.rio.crs)
    da_clip = da.rio.clip(
        gdf_proj.geometry,
        gdf_proj.crs,
        drop=True,
        all_touched=True,
    )
    return da_clip


def _get_granule_date_str(granule):
    umm = granule["umm"]
    te = umm.get("TemporalExtent") or umm.get("TemporalCoverage") or {}
    date = te.get("SingleDateTime")
    if not date:
        rd = te.get("RangeDateTime") or {}
        date = rd.get("BeginningDateTime") or "n/a"
    return date


def preview_swir_b04(granule, geom_geojson, out_dir=None, idx=None):
    tif_path = download_swir_b04(granule, out_dir=out_dir)
    da_clip = clip_to_geom(tif_path, geom_geojson)

    arr = da_clip.squeeze().values
    vmin, vmax = np.nanpercentile(arr, [2, 98])

    fig, ax = plt.subplots(figsize=(5, 5))
    da_clip.squeeze().plot.imshow(
        ax=ax,
        vmin=vmin,
        vmax=vmax,
        yincrease=False,
    )

    date_str = _get_granule_date_str(granule)
    if idx is not None:
        title = f"AST_07XT SWIR B04 - idx={idx} - {date_str}"
    else:
        title = f"AST_07XT SWIR B04 - {date_str}"
    ax.set_title(title)
    plt.tight_layout()
    plt.show()

    return da_clip, tif_path


def search_ast07xt_for_geom(
    geom_geojson,
    temporal=("2000-01-01", "2015-01-01"),
    min_cover_ratio=0.99,
):
    """
    Busca AST_07XT pelo bounding_box, depois filtra:
      - footprint real (GPolygons) deve ter interseção com a folha
      - a razão área_intersecção / área_folha >= min_cover_ratio
      - o granule deve possuir banda SRF_SWIR_B04
    Retorna somente granules válidos e ordenados por nuvem.
    """
    folha_geom = shape(geom_geojson)
    minx, miny, maxx, maxy = folha_geom.bounds

    ea.login()

    results = ea.search_data(
        short_name="AST_07XT",
        version="004",
        bounding_box=(minx, miny, maxx, maxy),
        temporal=temporal,
    )

    filtered = []
    for g in results:
        if not _has_swir_b04(g):
            continue
        fp = _get_granule_footprint(g)
        if fp is None:
            continue
        if not fp.intersects(folha_geom):
            continue
        inter = fp.intersection(folha_geom)
        if inter.is_empty or folha_geom.area == 0:
            continue
        ratio = inter.area / folha_geom.area
        if ratio < min_cover_ratio:
            continue
        filtered.append(g)

    sorted_results = sort_by_cloud_ast(filtered)
    return filtered, sorted_results


# ======================================================================
# 2) Sentinel-2 L2A – nuvem, ordenação, preview B04
# ======================================================================

PC_STAC = "https://planetarycomputer.microsoft.com/api/stac/v1"


def search_s2_l2a_for_geom(
    geom_geojson,
    temporal=("2017-01-01", "2025-12-31"),
    cloud_max=60.0,
    limit=100,
    min_cover_ratio=0.99,
):
    cat = Client.open(PC_STAC)
    dt = f"{temporal[0]}/{temporal[1]}"

    search = cat.search(
        collections=["sentinel-2-l2a"],
        intersects=geom_geojson,
        datetime=dt,
        query={"eo:cloud_cover": {"lt": cloud_max}},
        limit=limit,
    )
    items_all = list(search.get_items())

    folha_geom = shape(geom_geojson)
    items_cov = []
    for it in items_all:
        if it.geometry is None:
            continue
        it_geom = shape(it.geometry)
        if not it_geom.intersects(folha_geom):
            continue
        inter = it_geom.intersection(folha_geom)
        if inter.is_empty or folha_geom.area == 0:
            continue
        ratio = inter.area / folha_geom.area
        if ratio < min_cover_ratio:
            continue
        items_cov.append(it)

    items_sorted = sorted(
        items_cov,
        key=lambda it: it.properties.get("eo:cloud_cover", 999.0),
    )
    return items_cov, items_sorted


def _get_s2_datetime(item):
    return item.properties.get("datetime", "n/a")


def open_clip_s2_band(item, band, geom_geojson):
    pc.sign_inplace(item)
    href = item.assets[band].href
    da = rxr.open_rasterio(href, masked=True)
    g = shape(geom_geojson)
    gdf = gpd.GeoDataFrame(geometry=[g], crs="EPSG:4326")
    gdf_proj = gdf.to_crs(da.rio.crs)
    da_clip = da.rio.clip(
        gdf_proj.geometry,
        gdf_proj.crs,
        drop=True,
        all_touched=True,
    )
    return da_clip


def preview_s2_b04(item, geom_geojson, idx=None):
    da_clip = open_clip_s2_band(item, "B04", geom_geojson)

    arr = da_clip.squeeze().values
    vmin, vmax = np.nanpercentile(arr, [2, 98])

    fig, ax = plt.subplots(figsize=(5, 5))
    da_clip.squeeze().plot.imshow(
        ax=ax,
        vmin=vmin,
        vmax=vmax,
        yincrease=False,
    )

    dt = _get_s2_datetime(item)
    cc = item.properties.get("eo:cloud_cover", None)
    if idx is not None:
        title = f"S2 L2A B04 - idx={idx} - {dt} - cloud≈{cc}"
    else:
        title = f"S2 L2A B04 - {dt} - cloud≈{cc}"
    ax.set_title(title)
    plt.tight_layout()
    plt.show()

    return da_clip


# ======================================================================
# 3) Exemplo de uso para uma folha
# ======================================================================

if __name__ == "__main__":
    from db_conn import get_folha_geom_geojson as folha

    folha_id = "SB21_ZA_II1_NE"
    geom = folha(folha_id)

    # ---------------------- AST_07XT ----------------------
    results_ast, sorted_ast = search_ast07xt_for_geom(
        geom,
        temporal=("2000-01-01", "2015-01-01"),
        min_cover_ratio=0.99,
    )
    print(f"Granules AST_07XT encontrados (com SWIR B04 e cobrindo a folha): {len(results_ast)}")

    if not sorted_ast:
        print("Nenhum granule AST_07XT válido encontrado.")
    else:
        N_ast = min(10, len(sorted_ast))
        for rank, (cc, gran, attr_name) in enumerate(sorted_ast[:N_ast]):
            date = _get_granule_date_str(gran)
            gran_id = gran["umm"].get("GranuleUR", "n/a")
            print(f"{rank:02d}  cloud≈{cc:5.1f}%  attr={attr_name}  date={date}  id={gran_id}")

        best_cc, best_gran, best_attr = sorted_ast[0]
        print("\nMelhor granule AST_07XT por nuvem (com SWIR B04):")
        print("cloud_mean≈", best_cc, "attr=", best_attr)
        da_ast, tif_ast = preview_swir_b04(best_gran, geom, idx=0)

    # ---------------------- Sentinel-2 L2A ----------------------
    items_s2, sorted_s2 = search_s2_l2a_for_geom(
        geom,
        temporal=("2017-01-01", "2025-12-31"),
        cloud_max=60.0,
        limit=100,
        min_cover_ratio=0.99,
    )
    print(f"\nItens Sentinel-2 L2A encontrados (cobrindo a folha): {len(items_s2)}")

    if not sorted_s2:
        print("Nenhuma cena Sentinel-2 L2A válida encontrada.")
    else:
        N_s2 = min(10, len(sorted_s2))
        for rank, it in enumerate(sorted_s2[:N_s2]):
            cc = it.properties.get("eo:cloud_cover", None)
            dt = _get_s2_datetime(it)
            print(f"{rank:02d}  cloud≈{cc:5.1f}%  datetime={dt}  id={it.id}")

        best_s2 = sorted_s2[0]
        print("\nMelhor cena S2 L2A por nuvem:")
        print(
            "cloud≈",
            best_s2.properties.get("eo:cloud_cover", None),
            "datetime=",
            _get_s2_datetime(best_s2),
            "id=",
            best_s2.id,
        )

        da_s2 = preview_s2_b04(best_s2, geom, idx=0)
```

```python
# -*- coding: utf-8 -*-
from __future__ import annotations

import re
import tempfile
from pathlib import Path

import earthaccess as ea
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import rioxarray as rxr
from shapely.geometry import shape, Polygon

from pystac_client import Client
import planetary_computer as pc


# ======================================================================
# AST_07XT – nuvem, footprint e SWIR B04
# ======================================================================

def _extract_numbers_from_attr(attr):
    vals = attr.get("Values") or []
    text = " ".join(str(v) for v in vals)
    nums_str = re.findall(r"[-+]?\d*\.?\d+(?:[eE][-+]?\d+)?", text)
    if not nums_str:
        return None
    return [float(s) for s in nums_str]


def get_cloud_from_umm(granule):
    umm = granule["umm"]
    attrs = umm.get("AdditionalAttributes", [])
    best = None
    for attr in attrs:
        name = (attr.get("Name") or "").lower()
        if "cloud" not in name:
            continue
        nums = _extract_numbers_from_attr(attr)
        if not nums:
            continue
        mean_val = float(np.mean(nums))
        if best is None:
            best = (mean_val, nums, name)
        else:
            if "cloud_coverage_data" in name and "cloud_coverage_data" not in best[2]:
                best = (mean_val, nums, name)
            elif mean_val < best[0]:
                best = (mean_val, nums, name)
    if best is None:
        return None, None, None
    return best


def sort_by_cloud_ast(results):
    tmp = []
    for g in results:
        cc_mean, cc_list, attr_name = get_cloud_from_umm(g)
        if cc_mean is None:
            cc_mean = 999.0
            attr_name = None
        tmp.append((cc_mean, g, attr_name))
    return sorted(tmp, key=lambda t: t[0])


def _get_granule_footprint(granule):
    # tenta via atributo earthaccess
    sc = getattr(granule, "spatial_coverage", None)
    if not sc:
        umm = granule["umm"]
        sc = (
            umm.get("SpatialExtent")
            or umm.get("SpatialCoverage")
            or {}
        )
    hor = sc.get("HorizontalSpatialDomain") or {}
    geom = hor.get("Geometry") or {}
    gpolys = geom.get("GPolygons") or []
    if not gpolys:
        return None
    boundary = gpolys[0].get("Boundary") or {}
    pts = boundary.get("Points") or []
    if not pts:
        return None
    coords = [(p["Longitude"], p["Latitude"]) for p in pts]
    return Polygon(coords)


def build_folha_coverage_table(geom_geojson, results):
    folha_geom = shape(geom_geojson)
    rows = []
    for g in results:
        fp = _get_granule_footprint(g)
        if fp is None:
            continue
        inter = fp.intersection(folha_geom)
        if inter.is_empty or folha_geom.area == 0.0:
            r_folha = 0.0
        else:
            r_folha = inter.area / folha_geom.area
        cc_mean, _, _ = get_cloud_from_umm(g)
        if cc_mean is None:
            cc_mean = 999.0
        umm = g["umm"]
        gran_id = umm.get("GranuleUR", "n/a")
        date = _get_granule_date_str(g)
        rows.append(
            {
                "granule_ur": gran_id,
                "cloud_mean": float(cc_mean),
                "r_folha": float(r_folha),
                "date": date,
                "obj": g,
            }
        )
    if not rows:
        return pd.DataFrame(columns=["granule_ur", "cloud_mean", "r_folha", "date", "obj"])
    df = pd.DataFrame(rows)
    df = df.sort_values(["r_folha", "cloud_mean"], ascending=[False, True]).reset_index(drop=True)
    return df


def download_swir_b04(granule, out_dir=None):
    out_dir = Path(out_dir or tempfile.mkdtemp(prefix="AST_07XT_"))
    files = ea.download([granule], local_path=out_dir)
    b4_list = [Path(f) for f in files if "SRF_SWIR_B04" in str(f)]
    if not b4_list:
        raise RuntimeError("Nenhum arquivo SRF_SWIR_B04* encontrado nesse granule.")
    return b4_list[0]


def clip_to_geom(tif_path, geom_geojson):
    g = shape(geom_geojson)
    gdf = gpd.GeoDataFrame(geometry=[g], crs="EPSG:4326")
    da = rxr.open_rasterio(tif_path, masked=True)
    gdf_proj = gdf.to_crs(da.rio.crs)
    da_clip = da.rio.clip(
        gdf_proj.geometry,
        gdf_proj.crs,
        drop=True,
        all_touched=True,
    )
    return da_clip


def _get_granule_date_str(granule):
    umm = granule["umm"]
    te = umm.get("TemporalExtent") or umm.get("TemporalCoverage") or {}
    date = te.get("SingleDateTime")
    if not date:
        rd = te.get("RangeDateTime") or {}
        date = rd.get("BeginningDateTime") or "n/a"
    return date


def preview_swir_b04(granule, geom_geojson, out_dir=None, idx=None):
    tif_path = download_swir_b04(granule, out_dir=out_dir)
    da_clip = clip_to_geom(tif_path, geom_geojson)
    arr = da_clip.squeeze().values
    vmin, vmax = np.nanpercentile(arr, [2, 98])
    fig, ax = plt.subplots(figsize=(5, 5))
    da_clip.squeeze().plot.imshow(ax=ax, vmin=vmin, vmax=vmax)
    date_str = _get_granule_date_str(granule)
    if idx is not None:
        title = f"AST_07XT SWIR B04 - idx={idx} - {date_str}"
    else:
        title = f"AST_07XT SWIR B04 - {date_str}"
    ax.set_title(title)
    plt.tight_layout()
    plt.show()
    return da_clip, tif_path


def search_ast07xt_for_geom(
    geom_geojson,
    temporal=("2000-01-01", "2015-01-01"),
    min_cover_ratio=0.0,
):
    g = shape(geom_geojson)
    minx, miny, maxx, maxy = g.bounds
    ea.login()
    results = ea.search_data(
        short_name="AST_07XT",
        version="004",
        bounding_box=(minx, miny, maxx, maxy),
        temporal=temporal,
    )
    # garante SWIR B04 disponível
    with_b4 = []
    for gr in results:
        if any("SRF_SWIR_B04" in link for link in gr.data_links()):
            with_b4.append(gr)
    if min_cover_ratio > 0.0:
        folha_geom = shape(geom_geojson)
        filtered = []
        for gr in with_b4:
            fp = _get_granule_footprint(gr)
            if fp is None or folha_geom.area == 0.0:
                continue
            inter = fp.intersection(folha_geom)
            r_folha = 0.0 if inter.is_empty else inter.area / folha_geom.area
            if r_folha >= min_cover_ratio:
                filtered.append(gr)
        with_b4 = filtered
    sorted_ast = sort_by_cloud_ast(with_b4)
    return with_b4, sorted_ast


# ======================================================================
# Sentinel-2 L2A
# ======================================================================

PC_STAC = "https://planetarycomputer.microsoft.com/api/stac/v1"


def search_s2_l2a_for_geom(
    geom_geojson,
    temporal=("2017-01-01", "2025-12-31"),
    cloud_max=60.0,
    limit=100,
    min_cover_ratio=0.0,
):
    cat = Client.open(PC_STAC)
    dt = f"{temporal[0]}/{temporal[1]}"
    search = cat.search(
        collections=["sentinel-2-l2a"],
        intersects=geom_geojson,
        datetime=dt,
        query={"eo:cloud_cover": {"lt": cloud_max}},
        limit=limit,
    )
    items = list(search.get_items())
    folha_geom = shape(geom_geojson)
    valid = []
    for it in items:
        geom_it = shape(it.geometry)
        inter = geom_it.intersection(folha_geom)
        if folha_geom.area == 0.0:
            r_folha = 0.0
        else:
            r_folha = 0.0 if inter.is_empty else inter.area / folha_geom.area
        if r_folha < min_cover_ratio:
            continue
        valid.append(it)
    items_sorted = sorted(
        valid,
        key=lambda it: it.properties.get("eo:cloud_cover", 999.0),
    )
    return valid, items_sorted


def _get_s2_datetime(item):
    return item.properties.get("datetime", "n/a")


def open_clip_s2_band(item, band, geom_geojson):
    pc.sign_inplace(item)
    href = item.assets[band].href
    da = rxr.open_rasterio(href, masked=True)
    g = shape(geom_geojson)
    gdf = gpd.GeoDataFrame(geometry=[g], crs="EPSG:4326")
    gdf_proj = gdf.to_crs(da.rio.crs)
    da_clip = da.rio.clip(
        gdf_proj.geometry,
        gdf_proj.crs,
        drop=True,
        all_touched=True,
    )
    return da_clip


def preview_s2_b04(item, geom_geojson, idx=None):
    da_clip = open_clip_s2_band(item, "B04", geom_geojson)
    arr = da_clip.squeeze().values
    vmin, vmax = np.nanpercentile(arr, [2, 98])
    fig, ax = plt.subplots(figsize=(5, 5))
    da_clip.squeeze().plot.imshow(ax=ax, vmin=vmin, vmax=vmax)
    dt = _get_s2_datetime(item)
    cc = item.properties.get("eo:cloud_cover", None)
    if idx is not None:
        title = f"S2 L2A B04 - idx={idx} - {dt} - cloud≈{cc}"
    else:
        title = f"S2 L2A B04 - {dt} - cloud≈{cc}"
    ax.set_title(title)
    plt.tight_layout()
    plt.show()
    return da_clip


# ======================================================================
# main – integra lógica antiga + nova cobertura da folha
# ======================================================================

def main(folha_id: str = "SB21_ZA_II1_NE"):
    from db_conn import get_folha_geom_geojson as folha

    geom = folha(folha_id)

    # ---------------------- AST_07XT ----------------------
    results_ast, sorted_ast = search_ast07xt_for_geom(
        geom,
        temporal=("2000-01-01", "2015-01-01"),
        min_cover_ratio=0.0,  # não filtra aqui; filtra via r_folha
    )
    print(
        f"Granules AST_07XT encontrados (com SWIR B04 e intersectando a folha): "
        f"{len(results_ast)}"
    )

    df_cov = build_folha_coverage_table(geom, results_ast)

    tol = 1e-3
    df_full = df_cov[df_cov["r_folha"] >= 1.0 - tol].copy()

    print("\nGranules com r_folha ≈ 1.0 (cobrindo ~100% da folha):")
    if df_full.empty:
        print("  nenhum.")
    else:
        print(
            df_full[["granule_ur", "cloud_mean", "r_folha", "date"]]
            .sort_values(["cloud_mean", "date"])
            .to_string(index=False)
        )

    print("\nTop 10 por r_folha (independente de ser exatamente 1.0):")
    if df_cov.empty:
        print("  nenhum.")
    else:
        print(
            df_cov[["granule_ur", "cloud_mean", "r_folha", "date"]]
            .head(10)
            .to_string(index=False)
        )

    best_ast = None
    if not df_full.empty:
        best_row = df_full.sort_values(["cloud_mean", "date"]).iloc[0]
        best_ast = best_row["obj"]
        print("\nMelhor granule FULL COVER:")
        print(best_row[["granule_ur", "cloud_mean", "r_folha", "date"]])
    elif sorted_ast:
        best_cc, best_ast, best_attr = sorted_ast[0]
        print("\nNenhum FULL COVER; usando melhor por nuvem:")
        print(
            "cloud_mean≈",
            best_cc,
            "id=",
            best_ast["umm"].get("GranuleUR", "n/a"),
            "date=",
            _get_granule_date_str(best_ast),
        )

    if best_ast is not None:
        da_ast, tif_ast = preview_swir_b04(best_ast, geom, idx=0)
    else:
        da_ast, tif_ast = None, None

    # ---------------------- Sentinel-2 L2A ----------------------
    items_s2, sorted_s2 = search_s2_l2a_for_geom(
        geom,
        temporal=("2000-01-01", "2017-12-31"),
        cloud_max=60.0,
        limit=100,
        min_cover_ratio=0.99,
    )
    print(f"\nItens Sentinel-2 L2A encontrados (cobrindo a folha): {len(items_s2)}")

    if not sorted_s2:
        print("Nenhuma cena Sentinel-2 L2A válida encontrada.")
        da_s2 = None
    else:
        N_s2 = min(10, len(sorted_s2))
        for rank, it in enumerate(sorted_s2[:N_s2]):
            cc = it.properties.get("eo:cloud_cover", None)
            dt = _get_s2_datetime(it)
            print(
                f"{rank:02d}  cloud≈{cc:5.1f}%  "
                f"datetime={dt}  id={it.id}"
            )
        best_s2 = sorted_s2[0]
        print("\nMelhor cena S2 L2A por nuvem:")
        print(
            "cloud≈",
            best_s2.properties.get("eo:cloud_cover", None),
            "datetime=",
            _get_s2_datetime(best_s2),
            "id=",
            best_s2.id,
        )
        da_s2 = preview_s2_b04(best_s2, geom, idx=0)

    return results_ast, sorted_ast, df_cov, df_full, items_s2, sorted_s2, da_ast, da_s2


if __name__ == "__main__":
    main(folha_id="SB21_ZA_II1_NE")
```
