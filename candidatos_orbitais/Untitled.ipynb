```python
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# ----------------------------------------------------------------------
# 1) Ler os CSVs de candidatos ASTER e Sentinel-2
# ----------------------------------------------------------------------
csv_path_aster = "aster_sb21_za_ii_25k_candidates.csv"
csv_path_s2    = "s2_sb21_za_ii_25k_candidates.csv"

df_aster = pd.read_csv(csv_path_aster)
df_s2    = pd.read_csv(csv_path_s2)

# Garantir tipos numéricos coerentes
for df in (df_aster, df_s2):
    for col in ["delta_days", "coverage_fraction", "coverage_percent", "cloud_cover"]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")
    if "datetime" in df.columns:
        df["datetime"] = pd.to_datetime(df["datetime"], errors="coerce")

# Lista de folhas presentes em qualquer um dos dois dataframes
folhas = sorted(
    set(df_aster["folha_codigo"].unique())
    | set(df_s2["folha_codigo"].unique())
)


# ----------------------------------------------------------------------
# 2) Função de plotagem: ASTER e Sentinel-2 para uma folha
# ----------------------------------------------------------------------
def plot_aster_s2_for_folha(folha_codigo: str):
    sub_a = df_aster[df_aster["folha_codigo"] == folha_codigo].copy()
    sub_s = df_s2[df_s2["folha_codigo"] == folha_codigo].copy()

    if sub_a.empty and sub_s.empty:
        print(f"Sem cenas ASTER nem Sentinel-2 para {folha_codigo}.")
        return

    # Figura com dois painéis: esquerda ASTER, direita Sentinel-2
    fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
    ax_a, ax_s = axes

    # -------------------------
    # ASTER
    # -------------------------
    if not sub_a.empty:
        sc_a = ax_a.scatter(
            sub_a["delta_days"],
            sub_a["coverage_fraction"],
            s=25,
            c=sub_a["cloud_cover"],
            cmap="viridis",
        )
        ax_a.set_title(f"ASTER – {folha_codigo}")
        ax_a.set_xlabel("|Δ dias| em relação à data alvo")
        ax_a.set_ylabel("fração de cobertura na folha")
        ax_a.grid(True)

        # Destacar o "melhor" ASTER (maior cobertura, depois menor nuvem, depois menor Δ dias)
        best_a = sub_a.sort_values(
            ["coverage_fraction", "cloud_cover", "delta_days"],
            ascending=[False, True, True],
        ).iloc[0]

        ax_a.scatter(
            best_a["delta_days"],
            best_a["coverage_fraction"],
            s=80,
            facecolors="none",
            edgecolors="black",
            linewidths=1.5,
        )
        ax_a.annotate(
            best_a["id"],
            (best_a["delta_days"], best_a["coverage_fraction"]),
            xytext=(5, 5),
            textcoords="offset points",
            fontsize=7,
            rotation=45,
        )

        cbar_a = fig.colorbar(sc_a, ax=ax_a)
        cbar_a.set_label("% nuvem (ASTER)")

        # Limites de eixo x/y mais agradáveis
        ax_a.set_ylim(0, 1.05 * sub_a["coverage_fraction"].max())
        ax_a.set_xlim(
            -1,
            1.05 * sub_a["delta_days"].max()
            if sub_a["delta_days"].notna().any()
            else 1,
        )
    else:
        ax_a.set_title(f"ASTER – {folha_codigo} (sem cenas)")
        ax_a.axis("off")

    # -------------------------
    # Sentinel-2
    # -------------------------
    if not sub_s.empty:
        sc_s = ax_s.scatter(
            sub_s["delta_days"],
            sub_s["coverage_fraction"],
            s=25,
            c=sub_s["cloud_cover"],
            cmap="viridis",
        )
        ax_s.set_title(f"Sentinel-2 – {folha_codigo}")
        ax_s.set_xlabel("|Δ dias| em relação à data de referência ASTER/SB")
        ax_s.grid(True)

        # Destacar o "melhor" Sentinel-2 (mesma regra)
        best_s = sub_s.sort_values(
            ["coverage_fraction", "cloud_cover", "delta_days"],
            ascending=[False, True, True],
        ).iloc[0]

        ax_s.scatter(
            best_s["delta_days"],
            best_s["coverage_fraction"],
            s=80,
            facecolors="none",
            edgecolors="black",
            linewidths=1.5,
        )
        ax_s.annotate(
            best_s["id"],
            (best_s["delta_days"], best_s["coverage_fraction"]),
            xytext=(5, 5),
            textcoords="offset points",
            fontsize=7,
            rotation=45,
        )

        cbar_s = fig.colorbar(sc_s, ax=ax_s)
        cbar_s.set_label("% nuvem (Sentinel-2)")

        ax_s.set_ylim(0, 1.05 * sub_s["coverage_fraction"].max())
        ax_s.set_xlim(
            -1,
            1.05 * sub_s["delta_days"].max()
            if sub_s["delta_days"].notna().any()
            else 1,
        )
    else:
        ax_s.set_title(f"Sentinel-2 – {folha_codigo} (sem cenas)")
        ax_s.axis("off")

    fig.suptitle(f"Folha {folha_codigo} – Cenas ASTER e Sentinel-2", y=1.02)
    plt.tight_layout()
    plt.show()


# ----------------------------------------------------------------------
# 3) UI interativa com Dropdown de folhas
# ----------------------------------------------------------------------
dropdown_folha = widgets.Dropdown(
    options=folhas,
    value=folhas[0] if len(folhas) > 0 else None,
    description="Folha:",
    layout=widgets.Layout(width="50%"),
)

out = widgets.Output()


def on_folha_change(change):
    if change["name"] == "value" and change["new"] is not None:
        with out:
            out.clear_output(wait=True)
            plot_aster_s2_for_folha(change["new"])


dropdown_folha.observe(on_folha_change, names="value")

display(dropdown_folha, out)

# desenha a primeira folha imediatamente
if dropdown_folha.value is not None:
    with out:
        out.clear_output(wait=True)
        plot_aster_s2_for_folha(dropdown_folha.value)
```
